```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import scienceplots
```

Question 2.5.4

```python
w = np.arange(0, 10, 0.01)
absZ = np.sqrt(1 + np.square(w - 1/w))

# Apply dark style
plt.style.use(['science', 'scatter', 'dark_background'])

# Setup figure
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlabel('$\omega$', color='white', fontsize=16)
ax.set_ylabel('$|Z|$', color='white', fontsize=16)

# Set tick params to white
ax.tick_params(colors='white')
ax.tick_params(labelsize=16)

# Remove grid and minor ticks
ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)
ax.grid(False)

# Set aspect ratio and limits
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)

# Plot roots as lines (continuous trace)
ax.plot(w, absZ, color='#00F5FF', linewidth=2, label="$|Z|$")
ax.plot(w, w, color='white', linestyle='--', label="$|Z| = \omega$")
ax.plot(w, 1/w, color='white', linestyle=':', label="$|Z| = 1/\omega$")

ax.legend(frameon=True, fontsize=16)
fig.savefig("chap2sec2.5ex4.pdf", format="pdf")
plt.show()
```

Question 2.5.7

```python
t = np.arange(0, 2*np.pi, 0.01)
y = np.sin(t)*np.sin(9*t)

# Apply dark style
plt.style.use(['science', 'scatter', 'dark_background'])

# Setup figure
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlabel('$t$', color='white', fontsize=16)
ax.set_ylabel('$y(t)$', color='white', fontsize=16)

# Set x ticks
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(['$0$', '$\pi/2$', '$\pi$', '$3\pi/2$', '$2\pi$'], color='white')

# Set tick params to white
ax.tick_params(colors='white')
ax.tick_params(labelsize=16)

# Remove grid and minor ticks
ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)
ax.grid(False)

# Set aspect ratio and limits
ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.5, 1.5)

# Plot roots as lines (continuous trace)
ax.plot(t, y, color='#39ff14', linewidth=2, label="$y(t)$")
ax.plot(t, 0*np.ones(len(t)), color='white', linestyle=':')

ax.legend(frameon=True, fontsize=16)
fig.savefig("chap2sec2.5ex7.pdf", format="pdf")
plt.show()
```

Question 2.5.16

```python
t = np.arange(0, 0.05, 0.0001)
R = [200, 912]
wn = 1e3

# Apply dark style
plt.style.use(['science', 'scatter', 'dark_background'])

# Setup figure
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlabel('$t$', color='white', fontsize=16)
ax.set_ylabel('$r(t)$', color='white', fontsize=16)

# Set x ticks

# Set tick params to white
ax.tick_params(colors='white')
ax.tick_params(labelsize=16)

# Remove grid and minor ticks
ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)
ax.grid(False)

# Set aspect ratio and limits
ax.set_xlim(0, 0.05)
ax.set_ylim(0, 2)

# colors and labels
colors = ['#00FFFF', '#FF073A']
labels = ['$p_1$', '$p_2$']

# Plot roots as lines (continuous trace)
for res, color, label in zip(R, colors, labels):
    wd = np.sqrt(np.square(wn) - (np.square(res/2)))
    r = 1 - wn/wd*np.exp(-res*t/2)*np.sin(wd*t + np.arcsin(wd/wn))
    ax.plot(t, r, color=color, linewidth=2, label=label)

ax.plot(t, 1.2*np.ones(len(t)), color='white', linestyle='--')
ax.plot(t, 1*np.ones(len(t)), color='white', linestyle='--')

ax.text(
    0.65, 1.3,
    "$r_{max} = 1.2$",
    transform=ax.get_yaxis_transform(),
    color="white",
    va="center",
    ha="center",
    fontsize=16
)

ax.text(
    0.65, 0.85,
    "$r_{\infty} = 1$",
    transform=ax.get_yaxis_transform(),
    color="white",
    va="center",
    ha="center",
    fontsize=16
)

ax.legend(frameon=True, fontsize=16)
fig.savefig("chap2sec2.5ex16.pdf", format="pdf")
plt.show()
```
