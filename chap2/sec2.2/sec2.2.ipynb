```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import scienceplots
```

Question 2.2.1

```python
# Define the complex numbers
s1 = 2 + 1j
s2 = 1 - 2j
sum_s = s1 + s2
diff_s = s1 - s2

# Apply dark style
plt.style.use(['science', 'scatter', 'dark_background'])

# Setup figure
fig, ax = plt.subplots(figsize=(8, 4))
ax.set_xlabel('Real', color='white', fontsize=16)
ax.set_ylabel('Imaginary', color='white', fontsize=16)

# Draw only axis lines at x=0 and y=0
ax.axhline(0, color='white', linewidth=1)
ax.axvline(0, color='white', linewidth=1)

# Remove grid
ax.grid(False)

# Set aspect ratio and limits
ax.set_xlim(-2, 5)
ax.set_ylim(-3, 4)

# Set tick params to white
ax.tick_params(colors='white')
ax.tick_params(labelsize=16)

# Plot arrows from origin to each complex number
def plot_vector(z, label):
    ax.arrow(0, 0, z.real, z.imag,
             head_width=0.15, head_length=0.25, 
             fc='white', ec='white', length_includes_head=True)
    ax.text(z.real + 0.2, z.imag + 0.2, label, fontsize=16, color='white')

# Plot each vector
plot_vector(s1, r'$s_1 = 2 + i$')
plot_vector(s2, r'$s_2 = 1 - 2i$')
plot_vector(sum_s, r'$s_1 + s_2 = 3 - i$')
plot_vector(diff_s, r'$s_1 - s_2 = 1 + 3i$')

ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)

# Save and show
fig.savefig("chap2sec2.2ex1.eps", format="eps")
plt.show()
```

Question 2.2.14

```python
# Original complex number
z = 2 + 1j

# (a) z
z_a = z

# (b) z^2
z_b = z**2

# (c) 1 / z
z_c = 1 / z

# (d) |z| (purely real)
z_d = abs(z) + 0j

points = [
    (z_a, r'$2 + i$'),
    (z_b, r'$(2 + i)^2$'),
    (z_c, r'$\frac{1}{2 + i}$'),
    (z_d, r'$|2 + i|$'),
]

# Style
plt.style.use(['science', 'scatter', 'dark_background'])

# Setup figure
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlabel('Real', color='white', fontsize=16)
ax.set_ylabel('Imaginary', color='white', fontsize=16)

# Draw only axis lines at x=0 and y=0
ax.axhline(0, color='white', linewidth=1)
ax.axvline(0, color='white', linewidth=1)

# Remove grid
ax.grid(False)

# Get plot limits
real_vals = [p.real for p, _ in points]
imag_vals = [p.imag for p, _ in points]
ax.set_xlim(0, max(real_vals) + 1)
ax.set_ylim(-1, max(imag_vals) + 1)

# Force integer tick labels
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
ax.yaxis.set_major_locator(MaxNLocator(integer=True))

# Set tick params to white
ax.tick_params(colors='white', labelsize=16)

# Colors for each vector
colors = ['C0', 'C1', 'C2', 'C3']

# Function to plot vectors from origin
def plot_vector(z, label, color):
    ax.arrow(
        0, 0, z.real, z.imag,
        head_width=0.15, head_length=0.25,
        fc=color, ec=color,
        length_includes_head=True,
        alpha=1.0, linewidth=2.0  
    )
    ax.text(z.real + 0.05, z.imag - 0.35, label, fontsize=20, color=color)

# Plot each vector
for (p, lbl), c in zip(points, colors):
    plot_vector(p, lbl, c)

ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)

# Save and show
fig.savefig("chap2sec2.2ex14.eps", format="eps")
plt.show()
```

Question 2.2.19

```python
# Parameters
t = np.linspace(0, 3.2, 500)  # time range

# Spirals
z1 = np.exp((1 - 1j)*t)
z2 = np.exp((2 - 2j)*t)

# Time T when z1 hits negative x-axis
T = np.pi  # approx 3.1415

# Position of z2 at time T
z2_at_T = np.exp((2 - 2j)*T)

# Find index of time closest to T
idx_T = np.abs(t - T).argmin()

# Plot style
plt.style.use(['science', 'scatter', 'dark_background'])

fig, ax = plt.subplots(figsize=(8, 8))

# Plot spirals
ax.plot(z1.real, z1.imag, label=r'$z_1(t) = e^{(1 - i)t}$', color='C0', zorder=15, linewidth=3)
ax.plot(z2.real, z2.imag, label=r'$z_2(t) = e^{(2 - 2i)t}$', color='C1', zorder=10, linewidth=3)

# Mark point where z1 hits negative x-axis
ax.plot(z1.real[idx_T], z1.imag[idx_T], 
        'o', color='white', markersize=10, label=r'$z_1(T), \; T = \pi$', zorder=20)

# Mark position of z2 at time T
ax.plot(z2_at_T.real, z2_at_T.imag, 'X', color='orange', markersize=10, label=r'$z_2(T)$', zorder=20)

# Annotate T and z2(T) with safe raw strings and %-formatting
annotation_text_T = r'$z_1(T) = -e^{\pi}, T = \pi$'
annotation_text_z2 = r'$z_2(T) = e^{2\pi}$'

arrow =ax.annotate(annotation_text_T, 
            xy=(z1.real[idx_T], z1.imag[idx_T]),
            xytext=(20, -30), textcoords='offset points',
            arrowprops=dict(arrowstyle="->", color='white', linewidth=2),
            color='white', fontsize=16)

arrow.set_zorder(20)

ax.annotate(annotation_text_z2, 
            xy=(z2_at_T.real, z2_at_T.imag), xytext=(-80, -20),
            textcoords='offset points', color='orange', fontsize=16)

# Axis formatting
ax.axhline(0, color='white', linewidth=1)
ax.axvline(0, color='white', linewidth=1)
ax.set_xlabel('Real', color='white', fontsize=16)
ax.set_ylabel('Imaginary', color='white', fontsize=16)
ax.grid(False)
ax.set_aspect('equal', adjustable='box')

# Limits a bit wider
all_reals = np.concatenate([z1.real, z2.real])
all_imags = np.concatenate([z1.imag, z2.imag])
ax.set_xlim(min(all_reals)-10, max(all_reals))
ax.set_ylim(min(all_imags)-10, max(all_imags)+10)

# Integer ticks without decimals
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
ax.yaxis.set_major_locator(MaxNLocator(integer=True))

# Tick colors
ax.tick_params(colors='white', labelsize=14)

# Disable minor ticks
ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)

ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), ncol=2, fontsize=20)

plt.title("Spirals $e^{(1 - i)t}$ and $e^{(2 - 2i)t}$", color='white', fontsize=18)

# Save and show
fig.savefig("chap2sec2.2ex19.eps", format="eps")
plt.show()
```
