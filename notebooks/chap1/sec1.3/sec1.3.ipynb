2. Starting from $y(0) = 1$, find the solution to $dy/dt = y$ at time $t = 1$. Starting from that $y(1)$, solve $dy/dt = −y$ to time $t = 2$. Draw a rough graph of $y(t)$ from $t = 0$ to $t = 2$. What does this say about $e^{-1}$ times $e$?

```python
t1 = np.arange(0,1,0.01)
t2 = np.arange(1,2,0.01)
y1 = np.exp(t1)
y2 = np.exp(2-t2)

plt.figure(figsize=(6,4))
plt.plot(t1,y1, linewidth=2, color=C.red)
plt.plot(t2,y2, linewidth=2, color=C.red)
plt.xticks(np.arange(0, 3, step=1), [0, 1, 2])
plt.xlim(0,2)
plt.ylim(0,)
plt.minorticks_off()
savefig(plt.gcf(), 'chap1sec1.3ex2.eps', bbox_inches=None)
plt.show()
```

3. Start with $y(0) = \$5000$. If this grows by $dy/dt = .02y$ until $t = 5$ and then jumps to $a = .04$ per year until $t = 10$, what is the account balance at $t = 10$?

```python
t1 = np.arange(0,5,0.01)
t2 = np.arange(5,10.01,0.01)

initial1 = 5000
initial2 = 5000*np.exp(0.02*5) / np.exp(0.04*5)

y1 = initial1*np.exp(0.02*t1)
y2 = initial2*np.exp(0.04*t2)

y2t10 = initial2*np.exp(0.04*10)
print(y2t10)

plt.figure(figsize=(6,4))
plt.plot(t1,y1, linewidth=2, color=C.magenta)
plt.plot(t2,y2, linewidth=2, color=C.magenta)
plt.xlim(0,10)
plt.ylim(4750,)
plt.minorticks_off()
savefig(plt.gcf(), 'chap1sec1.3ex3.eps', bbox_inches=None)
plt.show()
```

11. In a one-year investment of $y(0) = \$100$, suppose the interest rate jumps from $6\%$ to $10\%$ after six months. Does the equivalent rate for a whole year equal $8\%$, or more than $8\%$, or less than $8\%$?

```python
t1 = np.arange(0,6,0.01)
t2 = np.arange(6,12.01,0.01)
t3 = np.arange(0,12.01,0.01)

initial1 = 100
initial2 = 100*np.exp(0.06*6) / np.exp(0.1*6)

y1 = initial1*np.exp(0.06*t1)
y2 = initial2*np.exp(0.1*t2)
y3 = initial1*np.exp(0.08*t3)

y2t12 = initial2*np.exp(0.1*12)
print(y2t12)
y3t12 = initial1*np.exp(0.08*12)
print(y3t12)

plt.figure(figsize=(6,4))
plt.plot(t1,y1, linewidth=2, color=C.magenta)
plt.plot(t2,y2, linewidth=2, color=C.magenta, label='$a = 0.06, 0.10$')
plt.plot(t3,y3, linewidth=2, linestyle='dotted', color=C.violet, label='$a = 0.08$')
plt.xlim(0,12)
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.3ex11.eps', bbox_inches=None)
plt.show()
```

20. The most important curve in probability is the bell-shaped graph of $e^{−t/2}$. With a calculator or computer find this function at $t = −2, −1, 0, 1, 2$. Sketch the graph of $e^{−t/2}$ from $t = −\infty$ to $t = \infty$. _It never goes below zero_.

```python
t = np.arange(-5,5,0.01)
y = np.exp(-np.square(t)/2)

index = [-2,-1,0,1,2]
for i in index:
    print(np.exp(-np.square(i)/2))

plt.figure(figsize=(6,4))
plt.plot(t,y, linewidth=2, color=C.cyan)
plt.xlim(-5,5)
plt.ylim(0,)
plt.minorticks_off()
savefig(plt.gcf(), 'chap1sec1.3ex20.eps', bbox_inches=None)
plt.show()
```
