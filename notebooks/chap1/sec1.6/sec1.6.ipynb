9. Which solution ($y$ or $Y$) is eventually larger if $y(0) = 0$ and $Y(0) = 0$?

```python
t = np.arange(0,5.1,0.01)
y1 = 2*np.exp(t) - 2*(t - 1)
y2 = np.exp(2*t)/4 - (2*t + 1)/4

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.cyan, label="$y(t) = 2e^t - 2(t - 1)$")
plt.plot(t,y2, linewidth=2, color=C.magenta, label="$Y(t) = \\frac{e^{2t}}{4} - \\frac{2t + 1}{4}$")
plt.xlim(0,5)
plt.ylim(0,7000)
#plt.xticks(np.arange(-4,4.1,4), ['$-\\infty \leftarrow t$', '0', '$t \\rightarrow \\infty$'], fontsize=18)
plt.yticks(np.arange(0,7000,2000))
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=20)
savefig(plt.gcf(), 'chap1sec1.6ex9.eps', bbox_inches=None)
plt.show()
```
