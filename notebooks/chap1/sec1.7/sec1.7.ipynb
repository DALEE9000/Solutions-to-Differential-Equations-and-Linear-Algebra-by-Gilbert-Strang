1. If $y(0) = a/2b$, the halfway point on the $S$-curve is at $t = 0$. Show that $d = b$ and $y(t) = \frac{a}{de^{-at} + b} = \frac{a}{b} \frac{1}{e^{-at} + 1}$. Sketch the curve from $y_{-\infty} = 0$ to $y_{\infty} = a/b$.

```python
t = np.arange(-5,5,0.01)
a, b = 1, 1
y = (a/b)*(1/(np.exp(-a*t) + 1))

plt.figure(figsize=(6,4))
plt.plot(t,y, linewidth=2, color=C.cyan, label="$y(t) = \\frac{a}{b} \\frac{1}{e^{-at} + 1}$")
plt.scatter(0,1/2, s=40, facecolors=FG, edgecolors=FG, zorder=2, clip_on=False)
plt.text(1/2,1/2, '$y(0) = \\frac{a}{2b}$', fontsize=18)
plt.xlim(-5,5)
plt.ylim(0,1)
plt.xticks(np.arange(-4,4.1,4), ['$-\\infty \leftarrow t$', '0', '$t \\rightarrow \\infty$'], fontsize=18)
plt.yticks(np.arange(0,1.1,0.5), ['0', '$\\frac{a}{2b}$', '$\\frac{a}{b}$'], fontsize=18)
plt.hlines(1/2, xmin=-5, xmax=0, linestyle="dotted", linewidth=2, color=FG)
plt.vlines(0,ymin=0, ymax=1/2, linestyle="dotted", linewidth=2, color=FG)
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=20)
savefig(plt.gcf(), 'chap1sec1.7ex1.eps', bbox_inches=None)
plt.show()
```

8. (recommended) Show those 3 solutions to $y ′ = y − y^2$ in one graph! They start from $y(0) = 1/2$ and $2$ and $−1$. The $S$-curve climbs from $12$ to $1$. Above that, $y(t)$ descends from $2$ to $1$. Below the $S$-curve, $y(t)$ drops from $−1$ to $−\infty$. Can you see $3$ regions in the picture? Dropin curves above $y = 1$ and $S$-curves sandwiched between $0$ and $1$ and dropoff curves below $y = 0$.

```python
t = np.arange(-5,10,0.01)

t21 = np.arange(-5,-np.log(2),0.01)
t22 = np.arange(-np.log(2),10,0.01)

t31 = np.arange(-5,-np.log(1/2),0.01)
t32 = np.arange(-np.log(1/2),10,0.01)

y1 = 1/(np.exp(-t) + 1)

y21 = 1/((-1/2)*np.exp(-t21) + 1)
y22 = 1/((-1/2)*np.exp(-t22) + 1)

y31 = 1/(-2*np.exp(-t31) + 1)
y32 = 1/(-2*np.exp(-t32) + 1)

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.red, label="$y(t) = \\frac{1}{e^{-1} + 1}$")
plt.plot(t21,y21, linewidth=2, color=C.orange, label="$y(t) = \\frac{1}{(-1/2)e^{-1} + 1}$")
plt.plot(t22,y22, linewidth=2, color=C.orange)
plt.plot(t31,y31, linewidth=2, color=C.cyan, label="$y(t) = \\frac{1}{-2e^{-1} + 1}$")
plt.plot(t32,y32, linewidth=2, color=C.cyan)
plt.text(-4,2,'$y = 1$', fontsize=16)
plt.text(-4,-1.5,'$y = 0$', fontsize=16)
plt.hlines(1, xmin=-5, xmax=10, linestyle="dotted", linewidth=2, color=FG)
plt.hlines(0, xmin=-5, xmax=10, linestyle="dotted", linewidth=2, color=FG)
plt.xlim(-5,10)
plt.ylim(-8,8)

plt.minorticks_off()
plt.legend(loc= 'lower right', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.7ex8.eps', bbox_inches=None)
plt.show()
```

11. $y′ = y − y^2 − \frac{1}{4} = − (y − \frac{1}{2})^2$ shows critical harvesting with a double steady state at $y = Y = \frac{1}{2}$. The layer of $S$-curves shrinks to that single line. Sketch a dropin curve that starts above $y(0) = \frac{1}{2}$ and a dropoff curve that starts below $y(0) = \frac{1}{2}$.

```python
t = np.arange(-10,10,0.01)

t11 = np.arange(-10,-2,0.01)
t12 = np.arange(-2,10,0.01)

t21 = np.arange(-10,2,0.01)
t22 = np.arange(2,10,0.01)

y = np.ones(len(t))*(1/2)

y01 = 1
y11 = (y01 - 1/2)/(y01*t11 - t11/2 + 1) + 1/2
y12 = (y01 - 1/2)/(y01*t12 - t12/2 + 1) + 1/2

y02 = 0
y21 = (y02 - 1/2)/(y02*t21 - t21/2 + 1) + 1/2
y22 = (y02 - 1/2)/(y02*t22 - t22/2 + 1) + 1/2

plt.figure(figsize=(6,4))
plt.plot(t,y, linewidth=2, color=C.red, label="$y(0) = \\frac{1}{2}$")
plt.plot(t11,y11, linewidth=2, color=C.yellow, label="$y(0) = 1$")
plt.plot(t12,y12, linewidth=2, color=C.yellow)
plt.plot(t21,y21, linewidth=2, color=C.blue, label="$y(0) = 0$")
plt.plot(t22,y22, linewidth=2, color=C.blue)
plt.text(-7.5,1,'$y(t) = 1/2$', fontsize=16)
plt.xlim(-10,10)
plt.ylim(-5,5)

plt.minorticks_off()
plt.legend(loc= 'lower right', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.7ex11.eps', bbox_inches=None)
plt.show()
```

15. The threshold equation is the logistic equation backward in time:
$$
    -\frac{dy}{dt} = ay − by^2
$$
is the same as

$$
    \frac{dy}{dt} = −ay + by^2
$$
Now $Y = 0$ is the stable steady state. $Y = a/b$ is the unstable state (why?). If $y(0)$ is below the threshold $a/b$ then $y(t) → 0$ and the species
will die out.

Graph $y(t)$ with $y(0) < a/b$ (reverse $S$-curve). Then graph $y(t)$ with $y(0) > a/b$.

```python
t = np.arange(-5,5,0.01)

y01 = 0.5
y02 = 1.5
a = 1
b = 1
d1 = 1/y01 - b/a
d2 = 1/y02 - b/a

t1 = np.arange(-5,np.log(3),0.01)
t2 = np.arange(np.log(3),5,0.01)

y1 = 1/(d1*np.exp(a*t) + b)

y21 = 1/(d2*np.exp(a*t1) + b)
y22 = 1/(d2*np.exp(a*t2) + b)

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.yellow, label="$y(0) < a/b$")
plt.plot(t1,y21, linewidth=2, color=C.magenta, label="$y(0) > a/b$")
plt.plot(t2,y22, linewidth=2, color=C.magenta)
plt.text(2,1.1,'$y = a/b = 1$', fontsize=16)
plt.text(-3,0.1,'$y = 0$', fontsize=16)
plt.hlines(1, xmin=-5, xmax=10, linestyle="dotted", linewidth=2, color=FG)
plt.hlines(0, xmin=-5, xmax=10, linestyle="dotted", linewidth=2, color=FG)
plt.xlim(-5,5)
plt.ylim(-1,2)

plt.minorticks_off()
plt.legend(loc= 'lower left', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.7ex15.eps', bbox_inches=None)
plt.show()
```

17. (a) Find the steady states of the Gompertz equation $dy/dt = y(1 − \ln (y))$.

(b) Show that $z = \ln (y)$ satisfies the linear equation $dz/dt = 1 − z$.

(c) The solution $z(t) = 1 + e^{-t} (z(0) − 1)$ gives what formula for $y(t)$ from $y(0)$?

```python
t = np.arange(-3,7,0.01)

y01 = np.exp(1) - 1
y02 = np.exp(1)
y03 = np.exp(1) + 1

y1 = np.exp(1 + np.exp(-t)*(np.log(y01) - 1))
y2 = np.exp(1 + np.exp(-t)*(np.log(y02) - 1))
y3 = np.exp(1 + np.exp(-t)*(np.log(y03) - 1))

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.red, label="$y(0) < e$")
plt.plot(t,y2, linewidth=2, linestyle="dotted", color=FG, label="$y(0) = e$")
plt.plot(t,y3, linewidth=2, color=C.blue, label="$y(0) > e$")
plt.text(-2,3,'$y = e$', fontsize=16)
plt.xlim(-3,7)
plt.ylim(0,5)

plt.minorticks_off()
plt.legend(loc= 'lower right', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.7ex17.eps', bbox_inches=None)
plt.show()
```

20. $dy/dt = ay − y^3$ has how many steady states $Y$ for $a < 0$ and then $a > 0$? Graph those values $Y(a)$ to see a pitchfork bifurcation–new steady states suddenly appear as $a$ passes zero. The graph of $Y(a)$ looks like a pitchfork.

```python
a = np.arange(-5,10,0.01)

y1 = np.sqrt(a)
y2 = 0*np.ones(len(a))
y3 = -np.sqrt(a)

plt.figure(figsize=(6,4))
plt.plot(a,y1, linewidth=2, color=C.cyan)
plt.plot(a,y2, linewidth=2, color=C.cyan)
plt.plot(a,y3, linewidth=2, color=C.cyan)
plt.vlines(0,ymin=-5, ymax=5, linewidth=2, linestyle="dotted")
plt.text(5,3.5, "$Y(a) = \\sqrt{a}$", fontsize=16)
plt.text(5,-4, "$Y(a) = -\\sqrt{a}$", fontsize=16)
plt.xlim(-5,10)
plt.xlabel("$a$", fontsize=16)
plt.ylim(-5,5)
plt.ylabel("$Y(a)$", fontsize=16)

plt.minorticks_off()
savefig(plt.gcf(), 'chap1sec1.7ex20.eps', bbox_inches=None)
plt.show()
```
