1. Draw the graph of $y = e^t$ by hand, for $−1 \le t \le 1$. What is its slope $dy/dt$ at $t = 0$? Add the straight line graph of $y = et$. Where do those two graphs cross?

```python
t = np.arange(-1,1,0.01)
t1 = np.arange(-0.25,0.25,0.01)
y = np.exp(t)
y1 = np.exp(1)*t
y2 = t1 + 1
pointsx = [0,1]
pointsy = [1,np.exp(1)]

plt.figure(figsize=(6,4))
plt.plot(t,y, linewidth=2, color=C.violet, label='$y = e^t$')
plt.plot(t,y1, linewidth=2, linestyle='dotted', color=C.cyan, label='$y = et$')
plt.plot(t1,y2, linewidth=2, linestyle='dotted', color=C.blue)
plt.text(0,1.45,'$dy/dt = 1$', ha='center', va='center', fontsize=16)
plt.text(0.8,np.exp(1),'$(1, e)$', ha='center', va='center', fontsize=16)
plt.scatter(pointsx, pointsy, color=C.blue, linewidth=3, zorder=2)
plt.ylim(0,)
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.1ex1.eps', bbox_inches=None)
plt.show()
```

2. Draw the graph of $y_1 = e^{2t}$ on top of $y_2 = 2e^t$. Which function is larger at $t = 0$? Which function is larger at $t = 1$?

```python
t = np.arange(-1,1,0.01)
y1 = np.exp(2*t)
y2 = 2*np.exp(t)

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.violet, label='$y_1 = e^{2t}$')
plt.plot(t,y2, linewidth=2, color=C.cyan, label='$y_2 = 2e^t$')
plt.ylim(0,)
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.1ex2.eps', bbox_inches=None)
plt.show()
```

8. Why will the solution to $dy/dt = y^2$ grow faster than the solution to $dy/dt = y$ (if we start them both from $y = 1$ at $t = 0$)? The first solution blows up at $t = 1$. The second solution et grows exponentially fast but it never blows up.

```python
t = np.arange(0,1,0.01)
y1 = np.exp(t)
y2 = 1 / (1-t)

plt.figure(figsize=(6,4))
plt.plot(t,y1, linewidth=2, color=C.violet, label='$y_1 = e^t$')
plt.plot(t,y2, linewidth=2, color=C.cyan, label='$y_2 = 1 / (1-t)$')
plt.ylim(0,)
plt.minorticks_off()
plt.legend(loc= 'upper left', frameon=True, fontsize=16)
savefig(plt.gcf(), 'chap1sec1.1ex8.eps', bbox_inches=None)
plt.show()
```
