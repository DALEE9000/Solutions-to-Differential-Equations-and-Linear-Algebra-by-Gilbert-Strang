Question 2.1.3

```python
A = 1
omega = np.pi
T = 2

x_min, x_max = -0.1*T, 1.1*T
y_min, y_max = -2.5*A, 2*A

t = np.linspace(x_min, x_max, 500)
alpha1, alpha2 = np.pi / 3, -np.pi / 2
y1 = A * np.cos(omega * t - alpha1)
y2 = A * np.cos(omega * t - alpha2)


fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(t, y1, linewidth=2, color=C.cyan, label=r"$y(t) = A \cos(\omega t - \pi / 3)$")
ax.plot(t, y2, linewidth=2, color=C.red, label=r"$y(t) = A \cos(\omega t + \pi / 2)$")

ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)

ax.set_xticks([0, T])
ax.set_xticklabels([r"$t = 0$", r"$t = T$"], fontsize=18)

ax.set_yticks([-A, 0, A])
ax.set_yticklabels([r"$-A$", r"$0$", r"$A$"], fontsize=18)

ax.hlines(0, xmin=x_min, xmax=x_max, linestyle="solid", linewidth=1.5, color=FG, zorder=0)

for x_line in [0, T]:
    ax.vlines(x_line, ymin=y_min, ymax=y_max, linestyle="dotted", linewidth=1.5, color=FG)

x_min, x_max = x_min, x_max
y = 1.3*A
ax.hlines(y, x_min, x_max, linewidth=1.5, color=FG)

arrowprops = dict(arrowstyle='-|>', color=FG, linewidth=1.5)

ax.annotate('', xy=(T, y), xytext=(T - 0.1, y), arrowprops=arrowprops)

ax.annotate('', xy=(0, y), xytext=(0 + 0.1, y), arrowprops=arrowprops)

tick_height = 0.1
ax.plot([0, 0], [y - tick_height, y + tick_height], color=FG, linewidth=2.5)
ax.plot([T, T], [y - tick_height, y + tick_height], color=FG, linewidth=2.5)

ax.text(0.5*T, 1.5*A, r"$T = 2 \pi / \omega$", ha='center', fontsize=16, color=FG)

ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)

ax.legend(loc='lower center', frameon=True, fontsize=16)

savefig(fig, "chap2sec2.1ex3.eps", bbox_inches=None)
plt.show()
```
