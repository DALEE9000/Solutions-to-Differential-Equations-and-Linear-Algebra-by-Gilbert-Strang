```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import scienceplots

import sys; sys.path.insert(0, r"C:\Users\david\Documents\math")
from plotstyle import C, FG, BG, savefig, animation_style, saveanim, neon_cmap
```

Question 2.3.10

```python
# Define roots
B = np.arange(0, 100, 0.1)
s1 = (-B + np.sqrt(np.square(B) - 4 + 0j)) / 2
s2 = (-B - np.sqrt(np.square(B) - 4 + 0j)) / 2

# Apply dark style

# Setup figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlabel('Real', color=FG, fontsize=16)
ax.set_ylabel('Imaginary', color=FG, fontsize=16)

# Set tick params to white
ax.tick_params(colors=FG)
ax.tick_params(labelsize=16)

# Remove grid and minor ticks
ax.tick_params(which='minor', top=False, bottom=False, left=False, right=False)
ax.grid(False)

# Set aspect ratio and limits
ax.set_xlim(-2.2, 0.2)
ax.set_ylim(-1.2, 1.2)

# Plot roots as lines (continuous trace)
ax.plot(s1.real, s1.imag, color=C.cyan, label="s1")
ax.plot(s2.real, s2.imag, color=C.yellow, label="s2")

# Scatter points (optional: keep dots for emphasis)
ax.scatter(s1.real, s1.imag, s=50, color=C.cyan, alpha=0.5)
ax.scatter(s2.real, s2.imag, s=50, color=C.yellow, alpha=0.5)

# Add arrows at the end to indicate direction
arrow_scale = 15  # scale factor for arrow size

# s1 arrow
ax.annotate("",
            xy=(s1.real[-1], s1.imag[-1]),
            xytext=(s1.real[-2], s1.imag[-2]),
            arrowprops=dict(arrowstyle="->", color=C.cyan, lw=2))

# s2 arrow
ax.annotate("",
            xy=(s2.real[-1], s2.imag[-1]),
            xytext=(s2.real[-2], s2.imag[-2]),
            arrowprops=dict(arrowstyle="->", color=C.yellow, lw=2))

ax.legend(frameon=True, fontsize=16)
savefig(fig, "chap2sec2.3ex10.pdf", bbox_inches=None)
plt.show()
```
